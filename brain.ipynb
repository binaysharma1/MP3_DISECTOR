```python
import torch
from demucs.pretrained import get_model
from demucs.apply import apply_model

model = get_model("htdemucs")
model.eval()
```

```python
import torchaudio
%pip install torchcodec

waveform, sr = torchaudio.load("/home/zephyr/Projects/MP3_POST-MORTEM/audio/Teri Jhuki Nazar Official Lyrical Video - Murder 3  Pritam  Shafqat Amanat Ali  Bollywood Songs - Sony Music India.mp3")

waveform.shape, sr
```

```python
sources = apply_model(
    model,
    waveform.unsqueeze(0),
    device="cuda" if torch.cuda.is_available() else "cpu"
)
```

```python
vocals = sources[0, model.sources.index("vocals")]
```

```python
instrumental = (
    sources[0, model.sources.index("drums")]
    + sources[0, model.sources.index("bass")]
    + sources[0, model.sources.index("other")]
)
```

```python
torchaudio.save(
    "/home/zephyr/Projects/MP3_POST-MORTEM/audio-generated/vocals.mp3",
    vocals.cpu(),
    model.samplerate,
    format="mp3"
)

torchaudio.save(
    "/home/zephyr/Projects/MP3_POST-MORTEM/audio-generated/instrumental.mp3",
    instrumental.cpu(),
    model.samplerate,
    format="mp3"
)
```
